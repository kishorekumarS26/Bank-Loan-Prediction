Categorical Data Conversion

```python
for i in loan_dataset.columns:
  if loan_dataset[i].dtypes == 'object':
    print(i,'details:')
    print(loan_dataset[i].value_counts())
    print()
```

```python
loan_dataset = loan_dataset.replace({'Loan_Status':{'Y':0,'N':1}})
```

```python
loan_dataset = loan_dataset.replace({'Gender':{'Male':1,'Female':0},
                                     'Married':{'Yes':0,'No':1},
                                     'Education':{'Graduate':0,'Not Graduate':1},
                                     'Self_Employed':{'Yes':0,'No':1},
                                     'Property_Area':{'Semiurban':0,'Urban':1,'Rural':2},
                                     'Dependents':{'0':0,'1':1,'2':2,'3+':3}
                                     })
```

```python
loan_dataset
```

Removing Unwanted Column

```python
loan_dataset = loan_dataset.drop(columns = 'Loan_ID')
```

```python
loan_dataset
```

Feature Extraction

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,12))
corr_matrix = loan_dataset.corr()
sns.heatmap(corr_matrix, annot=True)
```

```python
loan_dataset
```

```python
loan_dataset = loan_dataset.drop(columns="Loan_Amount_Term")
```

```python
loan_dataset
```

```python
from sklearn.preprocessing import MinMaxScaler
import pickle
scaler = MinMaxScaler()
loan_dataset[['ApplicantIncome','CoapplicantIncome','LoanAmount']] = scaler.fit_transform(
    loan_dataset[['ApplicantIncome','CoapplicantIncome','LoanAmount']]
)
pickle.dump(scaler,open('scaler.pkl','wb'))
```

```python
loan_dataset
```

Loan Imbalance

```python
loan_dataset.Loan_Status.value_counts()
```

```python
x=loan_dataset.iloc[:0,-1]
y=loan_dataset.iloc[:,-1]
```

```python
x = loan_dataset.drop(columns=['Loan_Status'])
y = loan_dataset['Loan_Status']

from imblearn.combine import SMOTEENN

# Apply a combination of SMOTE and Edited Nearest Neighbors
smote_enn = SMOTEENN(random_state=42)

X_balanced, y_balanced = smote_enn.fit_resample(x, y)

y_balanced.value_counts()
```

SPLIT TRAIN AND TEST

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(X_balanced,y_balanced,test_size=0.1)
```

```python
print(xtrain.shape,ytrain.shape,xtest.shape,ytest.shape)
```

Machine Learning Model

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
model=model.fit(xtrain,ytrain)

```

TEST MODEL

```python
ypred=model.predict(xtest)
print(xtrain.shape,ytrain.shape,xtest.shape,ytest.shape)
```

VALIDATION

```python
from sklearn import metrics
```

```python
ACCURACY=metrics.accuracy_score(ytest,ypred)
ACCURACY
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(ytest, ypred)
print(cm)
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(ytest, ypred))
```

```python
from sklearn.svm import SVC
from sklearn import metrics

# SVM MODEL
svmmodel = SVC()

# TRAIN ML
SVM_model = svmmodel.fit(xtrain, ytrain)

# TEST MODEL
ypred = SVM_model.predict(xtest)

# ACCURACY
ACCURACY = metrics.accuracy_score(ytest, ypred)
print(f"Model Accuracy: {ACCURACY}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

# KNN MODEL
knmodel = KNeighborsClassifier(n_neighbors=10)

# TRAIN ML
KNN_model = knmodel.fit(xtrain, ytrain)

# TEST MODEL
ypred = KNN_model.predict(xtest)

# ACCURACY
ACCURACY = metrics.accuracy_score(ytest, ypred)
print(f"Model Accuracy: {ACCURACY}")
```
